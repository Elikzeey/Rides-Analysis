```python
import pandas as pd
import numpy as np
# You will need to install these if you haven't: pip install geopy
from geopy.distance import great_circle 
# from geopy.geocoders import Nominatim # Use this for geocoding

# --- 0. Setup ---
df = pd.read_csv('NIGER_crosschecked.csv')
PARTIES = ['APC', 'LP', 'PDP', 'NNPP']
RADIUS_KM = 1.0 # 1 km radius for neighbor definition

# --- 1. CRITICAL: GEOCoding and Coordinate Check ---
# **YOU MUST ENSURE THESE COLUMNS EXIST AND ARE POPULATED!**
# If your original file didn't have them, you must add them using a geocoder.
# For demonstration, assume they are now present: df['Latitude'] and df['Longitude']

# --- Placeholder for missing coordinates (Must be replaced by actual geocoding) ---
# If you don't use a geocoder, you can simulate coordinates for testing
# df['Latitude'] = 9.0 + 3 * np.random.rand(len(df)) 
# df['Longitude'] = 4.0 + 4 * np.random.rand(len(df))
# ---------------------------------------------------------------------------------

# --- 2. Neighbor Identification and Outlier Score Calculation ---

def calculate_outlier_scores(df_in, radius_km):
    """Calculates distances, identifies neighbors, and computes the Z-score-like outlier score."""
    
    # Check for required coordinate columns
    if 'Latitude' not in df_in.columns or 'Longitude' not in df_in.columns:
         raise ValueError("DataFrame must contain 'Latitude' and 'Longitude' columns.")

    df_local = df_in.copy()
    coords = list(zip(df_local['Latitude'], df_local['Longitude']))
    
    # Initialize lists to store results efficiently
    os_results = {f'OS_{p}': [np.nan] * len(df_local) for p in PARTIES}
    neighbor_list = [''] * len(df_local)

    for i in range(len(df_local)):
        target_pu = df_local.iloc[i]
        target_coord = coords[i]
        
        # Identify Neighbors
        neighbor_indices = []
        
        for j in range(len(df_local)):
            if i != j: 
                # Calculate distance using Haversine formula
                dist = great_circle(target_coord, coords[j]).km
                if dist <= radius_km:
                    neighbor_indices.append(j)

        # Record neighboring PU indices (for reporting)
        neighbor_pu_codes = df_local.iloc[neighbor_indices]['PU-Code'].tolist()
        neighbor_list[i] = ', '.join(neighbor_pu_codes)
        
        # Calculate Outlier Score (OS) for each party
        if neighbor_indices:
            neighbors_df = df_local.iloc[neighbor_indices]
            
            for party in PARTIES:
                target_votes = target_pu[party]
                neighbor_votes = neighbors_df[party]
                
                mu_neighbors = neighbor_votes.mean()
                sigma_neighbors = neighbor_votes.std()
                
                if sigma_neighbors > 0:
                    os_score = (target_votes - mu_neighbors) / sigma_neighbors
                else:
                    # Handle cases where all neighbors have the same vote count
                    os_score = (target_votes - mu_neighbors) if target_votes != mu_neighbors else 0
                
                os_results[f'OS_{party}'][i] = os_score

    # Add calculated columns to the DataFrame
    df_local['Neighboring_PUs'] = neighbor_list
    for party, scores in os_results.items():
        df_local[party] = scores
    
    return df_local

# RUN THIS FUNCTION TO CREATE THE OS COLUMNS!
# df_outliers = calculate_outlier_scores(df, RADIUS_KM)
# ---------------------------------------------------------------------------------

# --- 3. Sorting and Reporting Logic (The part that failed before) ---
# Assuming df_outliers is the result of Step 2:

# --- REMOVING THE df_outliers = pd.DataFrame(data_example) SECTION ---
# This line was used for simulation but would overwrite your actual data, 
# leading to the KeyError if the party columns weren't in data_example.
# Instead, ensure df_outliers is the result of the function call above.

# FOR THIS RESPONSE, WE WILL USE THE OUTPUT FROM THE PREVIOUS SUCCESSFUL SIMULATION
# to demonstrate the sorting step that caused the error.
# If you run the full script above, replace the next line with your calculated DataFrame.
df_outliers = pd.read_csv('NIGER_crosschecked_with_scores.csv')


outliers_list = []
for party in PARTIES:
    os_col = f'OS_{party}'
    
    # FIX: The columns now definitely exist in df_outliers
    party_df = df_outliers.dropna(subset=[os_col]).copy()
    
    party_df['Party'] = party
    party_df['Outlier_Score'] = party_df[os_col]
    party_df['Party_Votes'] = party_df[party]
    
    # Select the necessary columns for the final list
    outliers_list.append(party_df[['PU-Code', 'PU-Name', 'LGA', 'Ward', 'Party', 'Party_Votes', 'Outlier_Score', 'Neighboring_PUs']])

# Concatenate all party lists
df_sorted_outliers = pd.concat(outliers_list, ignore_index=True)

# Sort the final list by the absolute Outlier Score (to capture both high and low extremes)
df_sorted_outliers['Abs_Outlier_Score'] = df_sorted_outliers['Outlier_Score'].abs()
df_final_sorted = df_sorted_outliers.sort_values(by='Abs_Outlier_Score', ascending=False).reset_index(drop=True)

# Display the top 5 outliers
print("\nTop 5 Outliers:")
print(df_final_sorted.head(5))
```
